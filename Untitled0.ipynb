```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Generate time values (1000 points)
t = np.arange(0, 1000)

# Create SNR values using:
# base value + small wave + random noise
base_snr = 20
wave = 3 * np.sin(0.02 * t)
noise = np.random.normal(0, 1, size=len(t))

snr = base_snr + wave + noise

# Create dataset table
data = pd.DataFrame({"Time": t, "SNR": snr})

# Save the dataset to a CSV file
data.to_csv("snr_dataset.csv", index=False)

# Plot SNR vs Time
plt.figure(figsize=(10,4))
plt.plot(t, snr)
plt.xlabel("Time")
plt.ylabel("SNR (dB)")
plt.title("Generated SNR Dataset")
plt.grid(True)
plt.show()

data.head()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.neural_network import MLPRegressor
from sklearn.metrics import mean_squared_error

# Load the dataset
data = pd.read_csv("snr_dataset.csv")

# Create input (X) and output (y)
# X = SNR at time t
# y = SNR at time t+1
X = data["SNR"].values[:-1].reshape(-1,1)
y = data["SNR"].values[1:]

# Split into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, shuffle=False)

# Create a simple neural network model
model = MLPRegressor(hidden_layer_sizes=(16, 16), max_iter=500)

# Train the model
model.fit(X_train, y_train)

# Make predictions
y_pred = model.predict(X_test)

# Calculate error
mse = mean_squared_error(y_test, y_pred)
print("Mean Squared Error:", mse)

# Plot actual vs predicted SNR
plt.figure(figsize=(10,4))
plt.plot(y_test, label="Actual SNR")
plt.plot(y_pred, label="Predicted SNR")
plt.title("Actual vs Predicted SNR")
plt.xlabel("Time")
plt.ylabel("SNR (dB)")
plt.legend()
plt.grid(True)
plt.show()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.neural_network import MLPRegressor
from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import mean_squared_error

# Load data
data = pd.read_csv("snr_dataset.csv")
snr = data["SNR"].values

# Parameters
window_size = 6 # number of past points to predict next
hidden_layers = (128, 64, 32)  # MLP hidden layers

# Create sliding window
X = []
y = []
for i in range(len(snr) - window_size):
    X.append(snr[i:i+window_size])
    y.append(snr[i+window_size])

X = np.array(X)
y = np.array(y).reshape(-1,1)

# Normalize data
scaler_X = MinMaxScaler()
scaler_y = MinMaxScaler()
X_scaled = scaler_X.fit_transform(X)
y_scaled = scaler_y.fit_transform(y).ravel()
# Train-test split
X_train, X_test, y_train, y_test = train_test_split(X_scaled, y_scaled, test_size=0.2, shuffle=False)

# MLP Model
model = MLPRegressor(
    hidden_layer_sizes=hidden_layers,
    max_iter=1000,
    learning_rate_init=0.001,
    solver='adam',
    random_state=42
)
model.fit(X_train, y_train)

# Predictions
y_pred_scaled = model.predict(X_test)
y_pred = scaler_y.inverse_transform(y_pred_scaled.reshape(-1,1)).ravel()
y_test_actual = scaler_y.inverse_transform(y_test.reshape(-1,1)).ravel()

# Error
mse = mean_squared_error(y_test_actual, y_pred)
print("New MSE (improved):", mse)

# Plot
plt.figure(figsize=(10,4))
plt.plot(y_test_actual, label="Actual SNR")
plt.plot(y_pred, label="Predicted SNR")
plt.title("Improved SNR Prediction using Windowing + Normalization + Bigger MLP")
plt.xlabel("Time")
plt.ylabel("SNR (dB)")
plt.legend()
plt.grid(True)
plt.show()
```
